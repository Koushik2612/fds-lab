```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)
```

```python
# Read the dataset
df = pd.read_csv('/content/Automobile_data (2).csv')

# Display first 5 rows
df.head()
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print(df.isnull().sum())

plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df.hist(figsize=(18,12), bins=20)
plt.suptitle("Histogram of Numerical Features")
plt.show()
```

```python
numerical_cols = df.select_dtypes(include=np.number).columns

plt.figure(figsize=(18,8))
sns.boxplot(data=df[numerical_cols])
plt.xticks(rotation=90)
plt.title("Box Plot of Numerical Features")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='fuel-type', data=df)
plt.title("Fuel Type Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='body-style', data=df)
plt.xticks(rotation=45)
plt.title("Body Style Distribution")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='horsepower',
                y='price',
                data=df)

plt.title("Horsepower vs Price")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='engine-size',
                y='price',
                data=df)

plt.title("Engine Size vs Price")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='body-style',
            y='price',
            data=df)

plt.xticks(rotation=45)
plt.title("Average Price by Body Style")
plt.show()
```
