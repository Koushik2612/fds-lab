```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Titanic-Dataset (3).csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()
```

```python

print("First 5 records:")
display(df.head())

print("Last 5 records:")
display(df.tail())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Fill missing Cabin values
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
passenger_data = df[["PassengerId", "Name", "Sex", "Age", "Pclass", "Survived"]]

display(passenger_data.head())
```

```python
df_sorted = df.sort_values(by="Age")

display(df_sorted[["Name", "Age", "Sex"]].head(10))
```

```python
df = df.rename(columns={
    "PassengerId": "Passenger_ID",
    "Pclass": "Passenger_Class"
})

print(df.columns)
```

```python
def age_group(age):
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)

display(df[["Name", "Age", "Age_Group"]].head(10))
```

```python
df["Family_Size"] = df["SibSp"] + df["Parch"] + 1

display(df[["Name", "SibSp", "Parch", "Family_Size"]].head())
```

```python
df["Alone"] = np.where(df["Family_Size"] == 1, "Yes", "No")

display(df[["Name", "Family_Size", "Alone"]].head(10))
```

```python
age_by_class = df.groupby("Passenger_Class")["Age"].mean()

print(age_by_class)
```

```python
survival_by_class = df.groupby("Passenger_Class")["Survived"].mean()

print(survival_by_class)
```

```python
survival_percentage = df.groupby("Passenger_Class")["Survived"].mean() * 100

print(survival_percentage)
```

```python
gender_survival = df.groupby("Sex")["Survived"].mean() * 100

print("Survival percentage by gender:")
print(gender_survival)
```

```python
df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})

print(df[["Sex"]].head())
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
df["Fare"] = df["Fare"].clip(
    lower=lower_limit,
    upper=upper_limit
)

print("Fare outliers handled using capping.")
```
